```python
num = int(input("Enter the number: "))

if num <= 1:
    print("Not a prime")
else:
    for i in range(2, num):
        if num % i == 0:
            print(f"{num} is not a prime number")
            break
    else:
        # Runs only if the loop never hits 'break'
        print(f"{num} is a prime number")
```
