```python
#Q21 Web server log analysis

import pandas as pd 
df = pd.read_csv("https://docs.google.com/spreadsheets/d/1ufTlDOSNjN6PS78VlV8fvyAQHF20yCVbU0pe_T1oTbQ/export?format=csv")
#counting the status codes of http requests . 
status = df["status"]
counts = {}
for code in status:
  if code in counts:
    counts[code] += 1
  else:
    counts[code] = 1
print(f"The number of requests \n{counts}")

#finding the requests taking more then 2 seconds. 
response_time = df["response_time"]
con = response_time > 2
slow_logs = df[con]
print(f"The slow queries \n{slow_logs}")
#print(slow_logs)
```

```python
#Q23 threshold CPU optimization 
servers = {
    "web-01": 72,
    "web-02": 91,
    "web-03": 45,
    "db-01": 84,
    "db-02": 96,
    "backup-01": 65,
    "cache-01": 81
}

total_usage = 0 # the total usage is 0 initally  

for server_name, usage in servers.items(): # iterates through the server where maps the data to server_name 
                                           #and usage to the utilization where items are used to split the data . 

    total_usage += usage # Increments (or)  adds the the usage to the total_usage . 

    if usage > 90:
        print(f"{server_name}: CRITICAL")

    elif usage >= 81:
        print(f"{server_name}: WARNING")


average_usage = int(total_usage / len(servers)) # calculates the avg utilization of the disk 

print (f"Average disk usage: {average_usage}%")
```
